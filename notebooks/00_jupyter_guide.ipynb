### 4.2. 코드 Interrupt

실행 중인 셀 :

- 셀 왼쪽이 `[*]`
- 오른쪽 위 커널 아이콘이 ● (채워진 원)

멈추는 방법 :

- 툴바의 **■** 버튼
- `Esc` 를 누른 뒤 **`I` 를 두 번** 빠르게 (Interrupt)

실행을 중단하면 `KeyboardInterrupt` 가 뜹니다. **에러가 아니라 정상적인 중단 신호**입니다.

![code interrupt](resources/00_jupyter_guide/code-interrupt.png)

```python
import time

for i in range(1, 21):
    print(f"프레임 처리 중... {i}/20")
    time.sleep(1)

print("끝까지 실행됨")
```

멈추지 않고 계속 `[*]` 인 채로 응답이 없다면 커널이 멈춘 것입니다.
`Kernel → Restart Kernel` 로 되살립니다.

### 4.3. 코드 셀 사용

#### 4.3.3. Magic Command

Jupyter 가 제공하는 특수 명령입니다.

`%` 는 코드 한 줄, `%%` 는 셀 전체에 적용됩니다.

| 명령 | 동작 |
|---|---|
| `%%time` | 이 셀의 실행 시간 측정 |
| `%matplotlib inline` | 그래프를 노트북 안에 그림 |
| `%whos` | 현재 커널에 살아 있는 변수 목록 |
| `%pwd`, `%cd` | 작업 디렉터리 확인·이동 |

```python
%%time

total = sum(i * i for i in range(2_000_000))
print(f"{total:,}")
```

```python
%whos
```

## 5. 커널

커널은 노트북 뒤에서 도는 파이썬 프로세스로, 
셀을 실행하면 **커널**이 그 코드를 처리합니다.


### 5.1. 규칙

1. **모든 셀이 같은 메모리를 공유합니다.**
2. **파일에 쓰인 순서가 아니라, 내가 실행한 순서대로** 반영됩니다.
3. 셀 왼쪽 `[3]` 같은 숫자는 **몇 번째로 실행했는지**를 뜻합니다.
    - `[*]` : 지금 실행 중
    - `[ ]` : 아직 안 돌린 셀
4. 결과가 설명과 다르면 **커널을 재시작하고 위에서부터** 다시 실행합니다.

```python
# 의도된 실패: a 는 아래 셀에서 정의
print(a)
```

```python
a = 3
b = 5

a + b
```

```python
print (a)
```

### 5.2. 커널 재시작

커널을 재시작하면 **변수가 전부 사라집니다.** 

상단바의 `Kernel` 메뉴를 통해 커널을 재시작할 수 있습니다.

| 메뉴 | 하는 일 |
|---|---|
| Restart Kernel | 메모리만 비움 |
| Restart Kernel and Clear All Outputs | 메모리 + 화면의 출력까지 비움 |
| Restart Kernel and Run All Cells | 재시작 후 처음부터 끝까지 자동 실행 |

## 7. 단축키

### 7.1. 명령 모드와 편집 모드

단축키는 **명령 모드**에서만 동작합니다. 편집 중(셀 테두리가 초록색)에는 그냥 글자로 입력됩니다.

| 키 | 동작 |
|---|---|
| `Esc` | 편집 모드 → 명령 모드 |
| `Enter` / 더블클릭 | 명령 모드 → 편집 모드 |

### 7.2. 자주 쓰는 단축키

| 키 | 동작 |
|---|---|
| `Shift + Enter` | 실행하고 다음 셀로 |
| `Ctrl + Enter` | 실행하고 제자리에 머무름 |
| `Alt + Enter` | 실행하고 아래에 새 셀 추가 |
| `A` / `B` | 위 / 아래에 셀 추가 |
| `D` `D` | 셀 삭제 |
| `Z` | 삭제 취소 |
| `M` / `Y` | 마크다운 / 코드 셀로 전환 |
| `I` `I` | 실행 중단 (Interrupt) |
| `0` `0` | 커널 재시작 |
| `Ctrl + S` | 저장 |
| `Shift + ↑` / `↓` | 여러 셀 선택 |
